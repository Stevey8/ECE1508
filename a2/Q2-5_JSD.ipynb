```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0.01, 0.99, 1000)

# KL divergence P||Q
kl = theta * np.log(theta / 0.3) + (1 - theta) * np.log((1 - theta) / 0.7)

# JSD
M1 = (theta + 0.3) / 2
M2 = (1 - theta + 0.7) / 2
jsd = 0.5 * (theta * np.log(theta / M1) + (1 - theta) * np.log((1 - theta) / M2)) + \
      0.5 * (0.3 * np.log(0.3 / M1) + 0.7 * np.log(0.7 / M2))

# compute minima
theta_star_kl = theta[np.argmin(kl)]
theta_star_jsd = theta[np.argmin(jsd)]
min_kl = min(kl)
min_jsd = min(jsd)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(theta, kl)
plt.xlabel(r'$\theta$')
plt.title(r'$D_{KL}(P\|Q)$')
plt.axvline(x=theta_star_kl, color='r', linestyle='--', label=rf'$\theta^* = {theta_star_kl:.2f}$')
plt.axhline(y=min_kl, color='orange', linestyle='--', label=rf'$D_{{KL}}(\theta^*) = {min_kl:.4f}$')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(theta, jsd, color='green')
plt.xlabel(r'$\theta$')
plt.title(r'$D_{{JSD}}(P\|Q)$')
plt.axvline(x=theta_star_jsd, color='r', linestyle='--', label=rf'$\theta^* = {theta_star_jsd:.2f}$')
plt.axhline(y=min_jsd, color='orange', linestyle='--', label=rf'$D_{{JSD}}(\theta^*) = {min_jsd:.4f}$')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

print(f"KL minimized at theta* = {theta_star_kl:.2f}, D_KL = {min_kl:.4f}")
print(f"JSD minimized at theta* = {theta_star_jsd:.2f}, D_JSD = {min_jsd:.4f}")
```
