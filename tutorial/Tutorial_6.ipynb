**RealNVP: Real-valued Non-Volume Preserving Flow**

RealNVP is a normalizing flow model: it learns an invertible mapping between data samples `x` and simple latent variables `z` drawn from a standard normal distribution. Because the mapping is invertible, the same model can both evaluate exact likelihoods and generate new samples.

The key building block is an affine coupling layer. Each layer keeps part of the input fixed with a binary mask, then uses that fixed part to predict a scale and shift for the remaining coordinates. Alternating the mask across layers lets every dimension eventually be transformed while keeping the inverse operation cheap.

Training uses maximum likelihood. For a data point `x`, the model computes `z = f(x)` and adds the log-determinant of the Jacobian to the standard normal log probability of `z`. Sampling runs the learned transformation backward: draw `z` from the base normal distribution, then compute `x = f^{-1}(z)`.

In the simple implementation below, RealNVP is trained on a 2D ring-shaped toy distribution so we can directly visualize whether generated samples match the target data.

```python
import math
import torch
import torch.nn as nn


class CouplingMLP(nn.Module):
    def __init__(self, dim, hidden_dim=128):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 2 * dim),
        )
        nn.init.zeros_(self.net[-1].weight)
        nn.init.zeros_(self.net[-1].bias)

    def forward(self, x):
        shift, log_scale = self.net(x).chunk(2, dim=-1)
        log_scale = 2.0 * torch.tanh(log_scale)
        return shift, log_scale


class AffineCouplingLayer(nn.Module):
    def __init__(self, dim, mask, hidden_dim=128):
        super().__init__()
        self.register_buffer("mask", mask)
        self.net = CouplingMLP(dim, hidden_dim)

    def forward(self, x):
        x_masked = x * self.mask
        shift, log_scale = self.net(x_masked)
        transform_mask = 1.0 - self.mask

        z = x_masked + transform_mask * (x * torch.exp(log_scale) + shift)
        log_det = (transform_mask * log_scale).sum(dim=-1)
        return z, log_det

    def inverse(self, z):
        z_masked = z * self.mask
        shift, log_scale = self.net(z_masked)
        transform_mask = 1.0 - self.mask

        x = z_masked + transform_mask * ((z - shift) * torch.exp(-log_scale))
        log_det = -(transform_mask * log_scale).sum(dim=-1)
        return x, log_det


class RealNVP(nn.Module):
    def __init__(self, dim, num_coupling_layers=6, hidden_dim=128):
        super().__init__()
        self.dim = dim
        base_mask = (torch.arange(dim) % 2).float()
        masks = [base_mask if i % 2 == 0 else 1.0 - base_mask for i in range(num_coupling_layers)]
        self.layers = nn.ModuleList(
            [AffineCouplingLayer(dim, mask, hidden_dim) for mask in masks]
        )

    def forward(self, x):
        z = x
        total_log_det = torch.zeros(x.shape[0], device=x.device)
        for layer in self.layers:
            z, log_det = layer(z)
            total_log_det = total_log_det + log_det
        return z, total_log_det

    def inverse(self, z):
        x = z
        total_log_det = torch.zeros(z.shape[0], device=z.device)
        for layer in reversed(self.layers):
            x, log_det = layer.inverse(x)
            total_log_det = total_log_det + log_det
        return x, total_log_det

    def log_prob(self, x):
        z, log_det = self.forward(x)
        base_log_prob = -0.5 * (z.pow(2) + math.log(2 * math.pi)).sum(dim=-1)
        return base_log_prob + log_det

    @torch.no_grad()
    def sample(self, num_samples, device=None):
        if device is None:
            device = next(self.parameters()).device
        z = torch.randn(num_samples, self.dim, device=device)
        x, _ = self.inverse(z)
        return x


def sample_2d_ring(batch_size, num_modes=8, radius=2.0, noise_std=0.08, device="cpu"):
    mode_ids = torch.randint(num_modes, (batch_size,), device=device)
    angles = 2 * math.pi * mode_ids.float() / num_modes
    centers = radius * torch.stack([torch.cos(angles), torch.sin(angles)], dim=-1)
    return centers + noise_std * torch.randn(batch_size, 2, device=device)


def train_real_nvp(model, steps=1000, batch_size=256, lr=1e-3, device=None):
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    losses = []

    for step in range(steps):
        x = sample_2d_ring(batch_size, device=device)
        loss = -model.log_prob(x).mean()

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        losses.append(loss.item())

    return losses


# Quick sanity check: forward followed by inverse should recover the input.
real_nvp = RealNVP(dim=2, num_coupling_layers=4, hidden_dim=64)
x = torch.randn(8, 2)
z, log_det = real_nvp(x)
x_reconstructed, inverse_log_det = real_nvp.inverse(z)

print("maximum reconstruction error:", (x - x_reconstructed).abs().max().item())
print("log_prob shape:", real_nvp.log_prob(x).shape)
```

```python
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.manual_seed(0)

real_nvp = RealNVP(dim=2, num_coupling_layers=6, hidden_dim=128)
losses = train_real_nvp(
    real_nvp,
    steps=1000,
    batch_size=256,
    lr=1e-3,
    device=device,
)

real_nvp.eval()
with torch.no_grad():
    real_points = sample_2d_ring(1000, device=device).cpu()
    generated_points = real_nvp.sample(1000, device=device).cpu()

print(f"initial loss: {losses[0]:.3f}")
print(f"final loss: {losses[-1]:.3f}")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].scatter(real_points[:, 0], real_points[:, 1], s=8, alpha=0.5)
axes[0].set_title("Real samples")

axes[1].scatter(generated_points[:, 0], generated_points[:, 1], s=8, alpha=0.5, color="tab:orange")
axes[1].set_title("RealNVP samples")

axes[2].plot(losses)
axes[2].set_title("Training loss")
axes[2].set_xlabel("step")
axes[2].set_ylabel("negative log likelihood")

for ax in axes[:2]:
    ax.set_aspect("equal")
    ax.set_xlim(-3.5, 3.5)
    ax.set_ylim(-3.5, 3.5)
    ax.grid(alpha=0.2)

axes[2].grid(alpha=0.2)
plt.tight_layout()
```

**GAN: Generative Adversarial Networks**

GANs are a type of generative models which are made of two adversarial networks. One model is called the generator which maps a latent representation to a sample of a target probability distribution. In practice, the generator models, take a normal noise as input and they are supposed to give a sample of a target probability distribution (e.g. 0-9 digits) in return.

In this tutorial, we will review the different blocks of GANs, WGANs. We will implement simple training loops for them and look at the results.

**Let's start from the generator model.**

 The job of generator is to produce a fake image resembling a valid sample of the target probability distribution. The input to this model is a random noise. Since the noise has a probabilistic nature, inputs are different every time we sample the noise distribution therefore we will have different outputs.


```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class Generator(nn.Module):
    def __init__(self, noise_dim, img_dim=1):
        super(Generator, self).__init__()
        self.noise_dim = noise_dim
        self.img_dim = img_dim

        self.linear = nn.Linear(noise_dim, 7 * 7 * 256)

        self.conv1 = nn.ConvTranspose2d(256, 128, kernel_size=5, stride=1, padding=2)  ## Look at Pytorch documentation to see how H_out and W_out is obtained: https://docs.pytorch.org/docs/stable/generated/torch.nn.ConvTranspose2d.html
        self.bn1 = nn.BatchNorm2d(128)

        self.conv2 = nn.ConvTranspose2d(128, 64, kernel_size=5, stride=2, padding=2, output_padding=1) ## https://docs.pytorch.org/docs/stable/generated/torch.nn.ConvTranspose2d.html
        self.bn2 = nn.BatchNorm2d(64)

        self.conv3 = nn.ConvTranspose2d(64, img_dim, kernel_size=5, stride=2, padding=2, output_padding=1) ## https://docs.pytorch.org/docs/stable/generated/torch.nn.ConvTranspose2d.html

    def forward(self, x):
        x = self.linear(x)                 # [batch_size, z_dim] → [batch_size, 7*7*256]
        x = x.view(-1, 256, 7, 7)          # reshape → [batch_size, 256, 7, 7]
        x = F.relu(x)

        x = F.relu(self.bn1(self.conv1(x)))     # conv1 + BN + ReLU
        x = F.relu(self.bn2(self.conv2(x)))     # conv2 + BN + ReLU

        x = torch.tanh(self.conv3(x))           # final conv + Tanh (no BN)

        return x
```

Example of usage

```python
# Example Usage
NOISE_DIM = 100
generator = Generator(noise_dim=NOISE_DIM)
noise = torch.randn(16, NOISE_DIM)  # Batch size of 16
fake_images = generator(noise)  # Output shape: (16, 1, 28, 28)
```

```python
import matplotlib.pyplot as plt
from torchvision.utils import make_grid

def plot_generated_images(generator, noise_dim, num_images=16, device='cpu'):
    generator.to(device)
    generator.eval()
    with torch.no_grad():
        noise = torch.randn(num_images, noise_dim).to(device)
        fake_images = generator(noise)

    # Normalize images to [0,1] range for visualization
    images = (fake_images + 1) / 2  # assuming output is from tanh [-1,1]

    grid = make_grid(images, nrow=4, padding=2)

    plt.figure(figsize=(8,8))
    plt.axis('off')
    plt.imshow(grid.permute(1, 2, 0).cpu().numpy(), cmap='gray')
    plt.show()

# Usage Example:
noise_dim = 100
generator = Generator(noise_dim=noise_dim)
plot_generated_images(generator, noise_dim=noise_dim, num_images=16)
```

The result is as expected since the model is just instantiated and no training has done so far.

**Discriminator Model**

The discriminator model must decide if the input is a real image or a fake image. Therefore, it is basically a binary classifier.

```python
class Discriminator(nn.Module):
    def __init__(self, img_channels=1):
        super(Discriminator, self).__init__()

        self.model = nn.Sequential(
            # Input: [batch_size, 1, 28, 28]
            nn.Conv2d(img_channels, 64, kernel_size=5, stride=2, padding=2),  # Output: [batch_size, 64, 14, 14]
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(64, 128, kernel_size=5, stride=2, padding=2),  # Output: [batch_size, 128, 7, 7]
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Flatten(),
            nn.Linear(128 * 7 * 7, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.model(x)


```

```python
# Example usage
discriminator = Discriminator()
img_sample = torch.randn(16, 1, 28, 28)
prediction = discriminator(img_sample)  # Output shape: [16, 1]
print("Predictions: ", [prediction.detach()> 0.5])
```

Now, it is time to implement training loop.

```python
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# Training loop
batch_size = 128
z_dim = 100
lr = 2e-4
num_epochs = 20
Disc_gen_ratio = 5

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5])
])

dataloader = DataLoader(
    datasets.MNIST(root='.', download=True, transform=transform),
    batch_size=batch_size,
    shuffle=True
)

generator = Generator(z_dim).to(device)
discriminator = Discriminator().to(device)

criterion = nn.BCELoss()
optimizer_G = torch.optim.Adam(generator.parameters(), lr=lr, betas=(0.5, 0.999))
optimizer_D = torch.optim.Adam(discriminator.parameters(), lr=lr, betas=(0.5, 0.999))

for epoch in range(num_epochs):
    for real_imgs, _ in dataloader:
        real_imgs = real_imgs.to(device)
        batch_size_current = real_imgs.size(0)

        # Train Discriminator multiple times
        for _ in range(Disc_gen_ratio):
            optimizer_D.zero_grad()
            z = torch.randn(batch_size_current, z_dim, device=device)
            fake_imgs = generator(z).detach()

            real_labels = torch.ones(batch_size_current, 1, device=device)
            fake_labels = torch.zeros(batch_size_current, 1, device=device)

            real_loss = criterion(discriminator(real_imgs), torch.ones(batch_size_current, 1, device=device))
            fake_loss = criterion(discriminator(fake_imgs), torch.zeros(batch_size_current, 1, device=device))

            d_loss = (real_loss + fake_loss) / 2
            d_loss.backward()
            optimizer_D.step()

        # Train Generator once
        optimizer_G.zero_grad()
        z = torch.randn(batch_size_current, z_dim, device=device)
        fake_imgs = generator(z)
        g_loss = criterion(discriminator(fake_imgs), real_labels)
        g_loss.backward()
        optimizer_G.step()

    print(f"Epoch [{epoch+1}/{num_epochs}] | D Loss: {d_loss.item():.4f} | G Loss: {g_loss.item():.4f}")

```

Let's take a look at the model's generated examples.

**Wasserstein GAN (WGAN)**

Instead of discriminator, here we have a critic function. The output of this model is a logit (no sigmoid applied). Therefore, it could be any real number. The higher the value is, it means it is closer to a real instance rather than a fake one.
Here we also use weight-clipping to impose Lipschitz continuity of discriminator function. You saw the theoretical necessity for this measure.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from torchvision.utils import make_grid
import matplotlib.pyplot as plt

# Define Generator
class Generator(nn.Module):
    def __init__(self, z_dim=100, img_channels=1):
        super(Generator, self).__init__()
        self.model = nn.Sequential(
            nn.Linear(z_dim, 7 * 7 * 256),
            nn.ReLU(True),
            nn.Unflatten(1, (256, 7, 7)),
            nn.ConvTranspose2d(256, 128, 5, 1, 2),
            nn.BatchNorm2d(128),
            nn.ReLU(True),
            nn.ConvTranspose2d(128, 64, 5, 2, 2, 1),
            nn.BatchNorm2d(64),
            nn.ReLU(True),
            nn.ConvTranspose2d(64, img_channels, 5, 2, 2, 1),
            nn.Tanh()
        )

    def forward(self, x):
        return self.model(x)


# Define Critic (WGAN version of Discriminator)
class Critic(nn.Module):
    def __init__(self, img_channels=1):
        super(Critic, self).__init__()
        self.model = nn.Sequential(
            nn.Conv2d(img_channels, 64, 5, 2, 2),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(64, 128, 5, 2, 2),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Flatten(),
            nn.Linear(128 * 7 * 7, 1)
        )

    def forward(self, x):
        return self.model(x)


# Training loop for WGAN
batch_size = 128
z_dim = 100
lr = 5e-5
num_epochs = 20
critic_iterations = 5
weight_clip = 0.01

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5])
])

dataloader = DataLoader(
    datasets.MNIST(root='.', download=True, transform=transform),
    batch_size=batch_size,
    shuffle=True
)

generator = Generator(z_dim).to(device)
critic = Critic().to(device)

optimizer_G = torch.optim.RMSprop(generator.parameters(), lr=lr)
optimizer_C = torch.optim.RMSprop(critic.parameters(), lr=lr)

for epoch in range(num_epochs):
    for i, (real_imgs, _) in enumerate(dataloader):
        real_imgs = real_imgs.to(device)
        batch_size_current = real_imgs.size(0)

        # Train Critic multiple times
        for _ in range(critic_iterations):
            z = torch.randn(batch_size_current, z_dim, device=device)
            fake_imgs = generator(z).detach()

            optimizer_C.zero_grad()
            critic_loss = -(torch.mean(critic(real_imgs)) - torch.mean(critic(fake_imgs)))
            critic_loss.backward()
            optimizer_C.step()

            # Weight clipping
            for p in critic.parameters():
                p.data.clamp_(-weight_clip, weight_clip)

        # Train Generator
        z = torch.randn(batch_size_current, z_dim, device=device)
        fake_imgs = generator(z)

        optimizer_G.zero_grad()
        generator_loss = -torch.mean(critic(fake_imgs))
        generator_loss.backward()
        optimizer_G.step()

    print(f"Epoch [{epoch+1}/{num_epochs}] | Critic Loss: {critic_loss.item():.4f} | Generator Loss: {generator_loss.item():.4f}")


# Generate and display 16 images
def show_generated_images(generator, z_dim, num_images=16):
    generator.eval()
    with torch.no_grad():
        z = torch.randn(num_images, z_dim).to(device)
        fake_imgs = generator(z)
        fake_imgs = (fake_imgs + 1) / 2  # Rescale from [-1, 1] to [0, 1]

    grid = make_grid(fake_imgs, nrow=4)
    plt.figure(figsize=(8, 8))
    plt.axis("off")
    plt.imshow(grid.permute(1, 2, 0).cpu().numpy(), cmap="gray")
    plt.show()


```
